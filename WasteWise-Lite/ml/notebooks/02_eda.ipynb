# 02 - Exploratory data analysis

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent / "src"))
import pandas as pd, numpy as np
import matplotlib.pyplot as plt
from preprocessing import load_daily_sales, load_batches, load_products
sales, batches, products = load_daily_sales(), load_batches(), load_products()
```

```python
sales.groupby('date').units_sold.sum().plot(figsize=(10,3), title='Total units sold per day'); plt.show()
```

```python
sales.groupby(sales.date.dt.dayofweek).units_sold.mean().plot.bar(title='Average units by day of week'); plt.show()
```

```python
print('Stock-out days:', sales.stockout_flag.sum(), '| waste units:', sales.waste_units.sum())
sales.groupby('category')[['units_sold','waste_units']].sum()
```
